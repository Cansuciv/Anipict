```python
csv_path = "/kaggle/working/runs/detect/yolov11n_keciler_detecion/results.csv"

# Dosyayı oku ve boş satırları atla
with open(csv_path, 'r', encoding='utf-8') as f:
    lines = [ln.strip() for ln in f if ln.strip()]

header = lines[0].split(",")

def idx(colname):
    try:
        return header.index(colname)
    except ValueError:
        return None

# Kolon indeksleri
i_epoch = idx("epoch")
i_prec  = idx("metrics/precision(B)")
i_rec   = idx("metrics/recall(B)")
i_map50 = idx("metrics/mAP50(B)")
i_map5095 = idx("metrics/mAP50-95(B)")
i_tbox = idx("train/box_loss")
i_tcls = idx("train/cls_loss")
i_tdfl = idx("train/dfl_loss")
i_vbox = idx("val/box_loss")
i_vcls = idx("val/cls_loss")
i_vdfl = idx("val/dfl_loss")

# Hesaplamalar
rows = []
for line in lines[1:]:
    parts = line.split(",")
    if len(parts) != len(header):
        continue

    def to_float_at(i):
        if i is None: 
            return None
        try:
            return float(parts[i])
        except:
            return None

    epoch = parts[i_epoch] if i_epoch is not None else None
    precision = to_float_at(i_prec)
    recall = to_float_at(i_rec)
    mAP50 = to_float_at(i_map50)
    mAP50_95 = to_float_at(i_map5095)
    tbox = to_float_at(i_tbox)
    tcls = to_float_at(i_tcls)
    tdfl = to_float_at(i_tdfl)
    vbox = to_float_at(i_vbox)
    vcls = to_float_at(i_vcls)
    vdfl = to_float_at(i_vdfl)

    f1 = 2*(precision*recall)/(precision+recall) if precision and recall and (precision+recall)!=0 else None

    train_loss = None if (tbox is None or tcls is None or tdfl is None) else (tbox + tcls + tdfl)
    val_loss = None if (vbox is None or vcls is None or vdfl is None) else (vbox + vcls + vdfl)

    rows.append({
        "epoch": epoch,
        "precision": precision,
        "recall": recall,
        "F1": f1,
        "mAP50": mAP50,
        "mAP50-95": mAP50_95,
        "train/box_loss": tbox,
        "train/cls_loss": tcls,
        "train/dfl_loss": tdfl,
        "train_loss": train_loss,
        "val/box_loss": vbox,
        "val/cls_loss": vcls,
        "val/dfl_loss": vdfl,
        "val_loss": val_loss
    })

# Tüm epoch değerlerini yazdır (opsiyonel)
# for r in rows:
    # print(f"Epoch {r['epoch']}: Precision={r['precision']}, Recall={r['recall']}, F1={r['F1']}, mAP50={r['mAP50']}, mAP50-95={r['mAP50-95']}")
    # print(f"  Train Losses -> box: {r['train/box_loss']}, cls: {r['train/cls_loss']}, dfl: {r['train/dfl_loss']}, total: {r['train_loss']}")
    # print(f"  Val Losses   -> box: {r['val/box_loss']}, cls: {r['val/cls_loss']}, dfl: {r['val/dfl_loss']}, total: {r['val_loss']}\n")

# ✅ Son epoch değerlerini alt alta yazdır
if rows:
    last = rows[-1]
    print("📊 Son Epoch Özet (Alt Alta):")
    print(f"Epoch           : {last['epoch']}")
    print(f"Precision       : {last['precision']}")
    print(f"Recall          : {last['recall']}")
    print(f"F1-score        : {last['F1']}")
    print(f"mAP@50          : {last['mAP50']}")
    print(f"mAP@50-95       : {last['mAP50-95']}")
    print(f"Train Box Loss  : {last['train/box_loss']}")
    print(f"Train Cls Loss  : {last['train/cls_loss']}")
    print(f"Train DFL Loss  : {last['train/dfl_loss']}")
    print(f"Total Train Loss: {last['train_loss']}")
    print(f"Val Box Loss    : {last['val/box_loss']}")
    print(f"Val Cls Loss    : {last['val/cls_loss']}")
    print(f"Val DFL Loss    : {last['val/dfl_loss']}")
    print(f"Total Val Loss  : {last['val_loss']}")
```
