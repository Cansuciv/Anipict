```python
import os
from collections import Counter

# Veri seti ana klasörü
base_path = '/kaggle/input/bocekler-final-datset'

splits = ['train', 'valid', 'test']
class_counts = {split: Counter() for split in splits}
total_counts = Counter()

# Her split için sayım yap
for split in splits:
    labels_dir = os.path.join(base_path, split, 'labels')
    if not os.path.exists(labels_dir):
        print(f"⚠️ {split} klasörü bulunamadı!")
        continue

    for label_file in os.listdir(labels_dir):
        if not label_file.endswith('.txt'):
            continue
        with open(os.path.join(labels_dir, label_file), 'r') as f:
            for line in f:
                if line.strip():
                    class_id = int(line.split()[0])
                    class_counts[split][class_id] += 1
                    total_counts[class_id] += 1

# Sonuçları yazdır
print("=== TRAIN / VALID / TEST SAYILARI ===")
for split in splits:
    print(f"\n📂 {split.upper()} klasörü:")
    for cls, count in sorted(class_counts[split].items()):
        print(f"  Sınıf {cls}: {count} örnek")

print("\n=== TOPLAM SAYILAR (TÜM SPLITLER) ===")
for cls, count in sorted(total_counts.items()):
    print(f"  Sınıf {cls}: {count} örnek")

print(f"\n🎯 Toplam örnek sayısı: {sum(total_counts.values())}")
```

```python
import os
from collections import Counter

# Veri seti ana klasörü
base_path = '/kaggle/input/bocekler-final-datset'

splits = ['train', 'valid', 'test']
class_counts = {split: Counter() for split in splits}
total_counts = Counter()

# Her split için sayım yap
for split in splits:
    labels_dir = os.path.join(base_path, split, 'labels')
    if not os.path.exists(labels_dir):
        print(f"⚠️ {split} klasörü bulunamadı!")
        continue

    for label_file in os.listdir(labels_dir):
        if not label_file.endswith('.txt'):
            continue
        with open(os.path.join(labels_dir, label_file), 'r') as f:
            for line in f:
                if line.strip():
                    class_id = int(line.split()[0])
                    class_counts[split][class_id] += 1
                    total_counts[class_id] += 1

# === SPLIT bazlı sınıf sayıları ===
print("=== TRAIN / VALID / TEST SINIF SAYILARI ===")
for split in splits:
    print(f"\n📂 {split.upper()} klasörü:")
    for cls, count in sorted(class_counts[split].items()):
        print(f"  Sınıf {cls}: {count} örnek")
    print(f"➡️  {split.upper()} toplam: {sum(class_counts[split].values())} örnek")

# === TOPLAM SINIF SAYILARI ===
print("\n=== TOPLAM (TÜM SPLITLER) ===")
for cls, count in sorted(total_counts.items()):
    print(f"  Sınıf {cls}: {count} örnek")

print(f"\n🎯 GENEL TOPLAM ÖRNEK SAYISI: {sum(total_counts.values())}")
```
